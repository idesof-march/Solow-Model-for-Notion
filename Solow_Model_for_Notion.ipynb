```python
import numpy as np
import plotly.graph_objects as go
import ipywidgets as widgets
from ipywidgets import interact

# Define the production function components (without labor, standard Cobb-Douglas)
def solow_model(s=0.3, delta=0.1, alpha=0.3, manual_k=1.5):
    # Calculate exact steady state k* analytically: s * k^alpha = delta * k -> k* = (s/delta)^(1/(1-alpha))
    k_star = (s / delta) ** (1 / (1 - alpha))

    # Generate capital vector for plotting
    k_max = max(k_star * 2, 5.0)
    k = np.linspace(0.01, k_max, 500)

    # Curves
    fk = k ** alpha
    sfk = s * fk
    dk = delta * k

    # Create the Plotly chart
    fig = go.Figure()

    # 1. Add sf(k) Curve
    fig.add_trace(go.Scatter(
        x=k, y=sfk, mode='lines', name='sf(k) - Savings/Investment',
        line=dict(color='#3b82f6', width=3),
        hovertemplate='Capital (k): %{x:.2f}<br>Savings: %{y:.2f}<extra></extra>'
    ))

    # 2. Add delta*k Line
    fig.add_trace(go.Scatter(
        x=k, y=dk, mode='lines', name='δk - Depreciation Line',
        line=dict(color='#ef4444', width=3),
        hovertemplate='Capital (k): %{x:.2f}<br>Depreciation: %{y:.2f}<extra></extra>'
    ))

    # 3. Add f(k) Production Function (Optional context curve)
    fig.add_trace(go.Scatter(
        x=k, y=fk, mode='lines', name='f(k) - Production Function',
        line=dict(color='#94a3b8', width=1.5, dash='dash'),
        hovertemplate='Capital (k): %{x:.2f}<br>Output: %{y:.2f}<extra></extra>'
    ))

    # 4. Add analytical steady-state vertical line (k*)
    fig.add_shape(type="line", x0=k_star, y0=0, x1=k_star, y1=s*(k_star**alpha)*1.2,
                  line=dict(color="#10b981", width=2, dash="dash"))

    fig.add_trace(go.Scatter(
        x=[k_star], y=[s * (k_star ** alpha)], mode='markers+text',
        name=f'Steady State k* ({k_star:.2f})',
        marker=dict(color='#10b981', size=10),
        text=[f"k* = {k_star:.2f}"], textposition="top left",
        hovertemplate='Equilibrium k*: %{x:.2f}<extra></extra>'
    ))

    # 5. Add user-controlled manual k slider line
    manual_sfk = s * (manual_k ** alpha)
    manual_dk = delta * manual_k
    fig.add_shape(type="line", x0=manual_k, y0=0, x1=manual_k, y1=max(manual_sfk, manual_dk)*1.2,
                  line=dict(color="#8b5cf6", width=2, dash="dot"))

    fig.add_trace(go.Scatter(
        x=[manual_k], y=[manual_sfk], mode='markers',
        name='Current Position on sf(k)',
        marker=dict(color='#8b5cf6', size=8, symbol='diamond'),
        hovertemplate='Your k: %{x:.2f}<br>Investment: %{y:.2f}<extra></extra>'
    ))

    # Layout styling matching a clean workspace UI
    fig.update_layout(
        title="Interactive Solow Growth Model",
        xaxis_title="k (Capital per Worker)",
        yaxis_title="Output / Investment / Depreciation",
        template="plotly_white",
        hovermode="x unified",
        legend=dict(yanchor="top", y=0.99, xanchor="left", x=0.01),
        xaxis=dict(range=[0, k_max]),
        yaxis=dict(range=[0, max(sfk)*1.3])
    )

    fig.show()

# Setup the sliders for the interactive controls
interact(
    solow_model,
    s=widgets.FloatSlider(value=0.30, min=0.05, max=0.95, step=0.05, description='Savings (s)'),
    delta=widgets.FloatSlider(value=0.10, min=0.02, max=0.50, step=0.01, description='Deprec. (δ)'),
    alpha=widgets.FloatSlider(value=0.30, min=0.10, max=0.90, step=0.05, description='Capital Share (α)'),
    manual_k=widgets.FloatSlider(value=2.0, min=0.1, max=10.0, step=0.1, description='Current k')
);
```

```python
import numpy as np
import plotly.graph_objects as go
import ipywidgets as widgets
from ipywidgets import interact, HTML

# Define the production function components (without labor, standard Cobb-Douglas)
def solow_model(s=0.3, delta=0.1, alpha=0.3, manual_k=2.0):
    # Calculate exact steady state k* analytically
    k_star = (s / delta) ** (1 / (1 - alpha))

    # Generate capital vector for plotting
    k_max = max(k_star * 2, 5.0)
    k = np.linspace(0.01, k_max, 500)

    # Mathematical Curves
    fk = k ** alpha
    sfk = s * fk
    dk = delta * k

    # Calculate Live Values at the current 'manual_k' positioning
    current_y = manual_k ** alpha
    current_i = s * current_y
    current_dk = delta * manual_k
    current_c = current_y - current_i # Consumption = Output - Investment

    # Create the Plotly chart
    fig = go.Figure()

    # 1. Add sf(k) Curve
    fig.add_trace(go.Scatter(
        x=k, y=sfk, mode='lines', name='sf(k) - Savings/Investment',
        line=dict(color='#3b82f6', width=3),
        hovertemplate='Capital (k): %{x:.2f}<br>Savings: %{y:.2f}<extra></extra>'
    ))

    # 2. Add delta*k Line
    fig.add_trace(go.Scatter(
        x=k, y=dk, mode='lines', name='δk - Depreciation Line',
        line=dict(color='#ef4444', width=3),
        hovertemplate='Capital (k): %{x:.2f}<br>Depreciation: %{y:.2f}<extra></extra>'
    ))

    # 3. Add f(k) Production Function
    fig.add_trace(go.Scatter(
        x=k, y=fk, mode='lines', name='f(k) - Production Function',
        line=dict(color='#94a3b8', width=1.5, dash='dash'),
        hovertemplate='Capital (k): %{x:.2f}<br>Output: %{y:.2f}<extra></extra>'
    ))

    # 4. Add analytical steady-state vertical line (k*)
    fig.add_shape(type="line", x0=k_star, y0=0, x1=k_star, y1=s*(k_star**alpha)*1.2,
                  line=dict(color="#10b981", width=2, dash="dash"))

    fig.add_trace(go.Scatter(
        x=[k_star], y=[s * (k_star ** alpha)], mode='markers+text',
        name=f'Steady State k* ({k_star:.2f})',
        marker=dict(color='#10b981', size=10),
        text=[f"k* = {k_star:.2f}"], textposition="top left",
        hovertemplate='Equilibrium k*: %{x:.2f}<extra></extra>'
    ))

    # 5. Add current position dot line tracker
    fig.add_shape(type="line", x0=manual_k, y0=0, x1=manual_k, y1=max(current_i, current_dk)*1.2,
                  line=dict(color="#8b5cf6", width=2, dash="dot"))

    fig.add_trace(go.Scatter(
        x=[manual_k], y=[current_i], mode='markers',
        name='Current Position',
        marker=dict(color='#8b5cf6', size=9, symbol='diamond'),
        hovertemplate='Your k: %{x:.2f}<br>Investment: %{y:.2f}<extra></extra>'
    ))

    # Layout styling matching a clean workspace UI
    fig.update_layout(
        title="Interactive Solow Growth Model Dashboard",
        xaxis_title="k (Capital per Worker)",
        yaxis_title="Output / Investment / Depreciation",
        template="plotly_white",
        hovermode="x unified",
        legend=dict(yanchor="top", y=0.99, xanchor="left", x=0.01),
        xaxis=dict(range=[0, k_max]),
        yaxis=dict(range=[0, max(sfk)*1.3]),
        margin=dict(l=40, r=40, t=40, b=40)
    )

    fig.show()

    # Print the clean dashboard panel underneath the chart
    print("\n" + "="*50)
    print(f" LIVE METRICS PANEL (At chosen Capital k = {manual_k:.2f})")
    print("="*50)
    print(f" Output per worker y=f(k):      {current_y:.4f}")
    print(f" Investment per worker i=sf(k): {current_i:.4f}")
    print(f" Depreciation δk:                {current_dk:.4f}")
    print(f" Consumption per worker c=(y-i): {current_c:.4f}")
    print("-"*50)
    if np.isclose(manual_k, k_star, atol=0.1):
        print("  STATUS: Economy is in Steady-State Equilibrium (i = δk)!")
    elif current_i > current_dk:
        print("  STATUS: Capital Accumulation occurring (i > δk)")
    else:
        print(" STATUS: Capital Depletion occurring (i < δk)")
    print("="*50)

# Setup the UI sliders
interact(
    solow_model,
    s=widgets.FloatSlider(value=0.30, min=0.05, max=0.95, step=0.05, description='Savings (s)'),
    delta=widgets.FloatSlider(value=0.10, min=0.02, max=0.50, step=0.01, description='Deprec. (δ)'),
    alpha=widgets.FloatSlider(value=0.30, min=0.10, max=0.90, step=0.05, description='Cap. Share (α)'),
    manual_k=widgets.FloatSlider(value=2.0, min=0.1, max=10.0, step=0.1, description='Your Cap. (k)')
);
```
